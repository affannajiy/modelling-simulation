## 2. Introduction

### What is a Python Generator?

A **generator** is a special function that produces values **one at a time** instead of all at once.

| Function type | Keyword used |
| :--- | :--- |
| Normal function | `return` |
| Generator | `yield` |

## 3. Why Use Generators?

Generators are useful when:

- data comes **step-by-step**
- events happen **over time**
- simulations run **continuously**

**Examples:** traffic lights, customer arrivals, machine operations, queue systems

### Key terms

| Term | Meaning |
| :--- | :--- |
| `yield` | give one value, then pause |
| `next()` | get the next value from a generator |
| generator | a function that produces values step-by-step |

## 4. Normal Function vs Generator

### Normal Function

```python
def numbers():
    return [1, 2, 3]

print(numbers())
```

**Output:**
```
[1, 2, 3]
```

### Generator Function

```python
def numbers():
    yield 1
    yield 2
    yield 3

g = numbers()
print(next(g))
print(next(g))
print(next(g))
```

**Output:**
```
1
2
3
```

````python
def numbers():
    yield 1
    yield 2
    yield 3
````

````python
g = numbers()

print(next(g))
print(next(g))
print(next(g))
````

## 5. Important Concept

> **`yield` pauses the function.**

Unlike `return`, `yield`:

1. **saves** the function state (its variables)
2. **continues** later from where it stopped
3. **produces** values one-by-one

### Example 1 - Simple Counter

````python
def counter():

    yield 1
    yield 2
    yield 3
````

````python
c = counter()

print(next(c))
print(next(c))
print(next(c))
````

### Example 2 - Simulation Clock

````python
def simulation_clock():
    time = 0
    while time < 5:
        yield time
        time += 1
````

````python
clock = simulation_clock()

for t in clock:
    print("Simulation Time:", t)
````

### Example 4 - Customer Arrival Simulation

````python
import random

def customer_arrival():

    customer = 1 #customer init with 1
    while customer <= 5:
        arrival_time = random.randint(1,60)
        yield f"Customer {customer} arrives after {arrival_time} minutes"
        customer += 1
````

````python
arrival = customer_arrival()

for a in arrival:
    print(a)
````

## 6. Activities

### Activity 1
> Create a generator that produces 10, 20, 30, 40 and 50.

**Explanation**

Firstly, a generator called `tens()` was created, where `n` starts at 10. Each `yield n` hands back the current value and pauses the function. When the next value is requested, it continues from where it stopped and adds 10 to `n`. Once `n` reaches 60, the `while` condition becomes false, so the function ends and the `for` loop stops on its own through `StopIteration`. This is similar to a bank queue number machine, which only prints the next number when the button is pressed.

**Code**

````python
def tens():
    n = 10
    while n <= 50:
        yield n      # pause here and hand n back
        n += 10      # resumes from here on the next request

for value in tens():
    print(value)
````

### Activity 2
> Create a weather generator that produces Sunny, Rainy and Cloudy, repeating continuously.

**Explanation**

Secondly, a weather generator was created using `while True`, which makes it an infinite generator, meaning it never runs out of values. It yields Sunny, Rainy and Cloudy, then starts again from Sunny, so the weather repeats continuously. The program does not freeze, because a generator only runs when the next value is requested. However, looping over it with `for x in weather():` would never end. Hence, `next(w)` was only called 9 times to display three full cycles, while the generator itself can continue forever.

**Code**

````python
def weather():
    while True:          # infinite generator, never ends
        yield "Sunny"
        yield "Rainy"
        yield "Cloudy"

w = weather()
for day in range(1, 10):     # we choose how many values to take
    print("Day", day, ":", next(w))
````

### Activity 3
> Create a machine operation generator that produces Machine Start, Machine Running and Machine Stop.

**Explanation**

Thirdly, a machine generator with exactly three `yield` statements was created, which makes it a finite generator. Each `next(m)` moves the machine from Start to Running to Stop, much like a washing machine cycle. A fourth `next(m)` raised `StopIteration`, which is the error Python uses to signal that a generator is finished. A `for` loop catches this error automatically, but calling `next()` by hand does not, so it was handled with a `try` block.

**Code**

````python
def machine():
    yield "Machine Start"
    yield "Machine Running"
    yield "Machine Stop"

m = machine()
print(next(m))
print(next(m))
print(next(m))

# A 4th next() has nothing left to give
try:
    print(next(m))
except StopIteration:
    print("Generator finished (StopIteration)")
````

## 7. Mini Challenge - Queue Simulation Generator

> Modify the code below:
>
> ```python
> import random
>
> def queue():
>     q = 0
>     for minute in range(10):
>         arrival = random.randint(0, 1)
>         if arrival == 1:
>             q += 1
>         yield q
> ```
>
> **Tasks:**
> 1. add customer service
> 2. print queue size
> 3. simulate 20 minutes

**Explanation**

The original code only adds customers, so the queue can only grow. Hence, customer service was added, whereby each minute one customer is served with a 50% chance, but only if `q > 0`, so the queue never becomes negative. The generator yields the minute, arrival, service and queue size, and runs for 20 minutes instead of 10. In this run, 11 customers arrived and 7 were served, leaving 4 in the queue, similar to a cafe at lunch hour. `random.seed(1)` keeps the result the same every run.

**Code**

````python
import random

def queue(minutes=20):
    q = 0
    for minute in range(1, minutes + 1):
        arrival = random.randint(0, 1)      # 1 = a customer arrives this minute
        if arrival == 1:
            q += 1

        served = 0
        if q > 0 and random.randint(0, 1) == 1:   # can only serve if someone is waiting
            q -= 1
            served = 1

        yield minute, arrival, served, q

random.seed(1)   # optional: fixes the random numbers so the result is repeatable

total_arrived = total_served = max_q = 0
print("Minute | Arrived | Served | Queue size")
for minute, arrived, served, size in queue(20):
    print(f"{minute:6} | {arrived:7} | {served:6} | {size:10}")
    total_arrived += arrived
    total_served += served
    max_q = max(max_q, size)

print("\nTotal arrived :", total_arrived)
print("Total served  :", total_served)
print("Longest queue :", max_q)
print("Left in queue :", total_arrived - total_served)
````

### Traffic Light System

**Explanation**

Firstly, the traffic light from Example 3 was improved by giving each light a duration, whereby RED lasts 5 seconds, GREEN 4 seconds and YELLOW 2 seconds. The outer `while True` loop repeats the cycle forever, while the inner loops count the seconds and yield the time and colour once per second. As a result, one cycle takes 11 seconds, so the light turns RED again at t = 11. This is closer to a real junction, as opposed to the lab example where each colour only lasts one step.

**Code**

````python
def traffic_light():
    durations = {"RED": 5, "GREEN": 4, "YELLOW": 2}   # seconds each light stays on
    t = 0
    while True:
        for colour in ["RED", "GREEN", "YELLOW"]:
            for _ in range(durations[colour]):
                yield t, colour
                t += 1

light = traffic_light()
for _ in range(15):
    t, colour = next(light)
    print(f"t = {t:2}s : {colour}")
````

### Elevator Simulation

**Explanation**

Secondly, an elevator for floors 0 to 5 was simulated with 4 random floor requests. For each request, it compares the target with its current floor to move up or down, one floor per time step. Every `yield` is one event, either moving one floor or opening the doors. In this run, the requests were floors 1, 4, 4 and 1, and the elevator finished at t = 11. However, it serves requests in order, so it does not take the shortest route like a real lift.

**Code**

````python
import random

def elevator(requests):
    floor = 0
    t = 0
    yield t, floor, "Idle at ground floor"
    for target in requests:
        while floor != target:
            direction = "Up" if target > floor else "Down"
            floor += 1 if direction == "Up" else -1
            t += 1
            yield t, floor, f"Moving {direction}"
        t += 1
        yield t, floor, f"Doors open at floor {floor}"

random.seed(3)
requests = [random.randint(0, 5) for _ in range(4)]
print("Requests:", requests)
for t, floor, event in elevator(requests):
    print(f"t = {t:2} | floor {floor} | {event}")
````

### ATM Queue

**Explanation**

Thirdly, an ATM queue with one machine was simulated for 8 customers, who arrive every 1 to 5 minutes and use the ATM for 2 to 4 minutes. The key line `start = max(clock, atm_free_at)` makes a customer start immediately if the ATM is free, or wait until it is free. The waiting time is the start time minus the arrival time. In this run, the average wait was 2.12 minutes, with customer 8 waiting the longest at 6 minutes. SimPy uses the same logic in future labs.

**Code**

````python
import random

def atm_queue(n_customers):
    clock = 0          # arrival time of the current customer
    atm_free_at = 0    # time when the ATM becomes free
    for c in range(1, n_customers + 1):
        clock += random.randint(1, 5)              # inter-arrival time
        start = max(clock, atm_free_at)            # wait if ATM is busy
        service = random.randint(2, 4)
        end = start + service
        atm_free_at = end
        yield c, clock, start, start - clock, end

random.seed(7)
waits = []
print("Customer | Arrive | Start | Wait | Leave")
for c, arrive, start, wait, leave in atm_queue(8):
    print(f"{c:8} | {arrive:6} | {start:5} | {wait:4} | {leave:5}")
    waits.append(wait)

print("\nAverage waiting time:", round(sum(waits) / len(waits), 2), "minutes")
````

### Weather System

Table 1 shows the chance of tomorrow's weather based on today's weather.

| Today / Tomorrow | Sunny | Cloudy | Rainy |
| :--- | :---: | :---: | :---: |
| **Sunny** | 0.6 | 0.3 | 0.1 |
| **Cloudy** | 0.3 | 0.4 | 0.3 |
| **Rainy** | 0.2 | 0.4 | 0.4 |

**Explanation**

Fourthly, the weather generator was made more realistic using a Markov chain, a model where the next state only depends on the current state. `random.choices(options, weights)` picks tomorrow's weather using the probabilities in Table 1, so a sunny day is most likely followed by another sunny day. Each row in the table adds up to 1.0. Since the generator is infinite, `next()` was called 30 times to simulate 30 days, which gave 16 sunny, 4 cloudy and 10 rainy days.

**Code**

````python
import random

def weather_system(start="Sunny"):
    # chance of tomorrow's weather, given today's weather
    transitions = {
        "Sunny":  {"Sunny": 0.6, "Cloudy": 0.3, "Rainy": 0.1},
        "Cloudy": {"Sunny": 0.3, "Cloudy": 0.4, "Rainy": 0.3},
        "Rainy":  {"Sunny": 0.2, "Cloudy": 0.4, "Rainy": 0.4},
    }
    today = start
    while True:
        yield today
        options = list(transitions[today].keys())
        weights = list(transitions[today].values())
        today = random.choices(options, weights)[0]

random.seed(5)
w = weather_system()
counts = {"Sunny": 0, "Cloudy": 0, "Rainy": 0}
for day in range(1, 31):
    today = next(w)
    counts[today] += 1
    if day <= 10:
        print("Day", day, ":", today)

print("\n30-day count:", counts)
````

### Dice Rolling Simulation

**Explanation**

Lastly, a fair die was rolled 10, 100 and 10,000 times, where each face should appear 1/6 of the time, which is 16.7%. With 10 rolls, face 4 appeared 40% of the time while face 6 never appeared. With 10,000 rolls, every face landed between 16.2% and 17.2%. This is the law of large numbers, where results get closer to the expected value as trials increase. Hence, simulations need many runs to be reliable, and the generator never stores all the rolls in memory.

**Code**

````python
import random

def dice_rolls(n):
    for _ in range(n):
        yield random.randint(1, 6)

random.seed(10)
for n in [10, 100, 10000]:
    counts = {face: 0 for face in range(1, 7)}
    for roll in dice_rolls(n):
        counts[roll] += 1
    print(f"{n} rolls:")
    for face in range(1, 7):
        print(f"  {face}: {counts[face]:5}  ({counts[face] / n * 100:.1f}%)")
````
